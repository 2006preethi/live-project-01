#TASK -02  DATA CLEANING

```python
df.columns= df.columns.str.strip().str.lower()
print(df.columns)
print(df["role"].unique())
```

```python
role_mapping ={"Backend Engineer": "SDE Backend","Backend Dev":"SDE Bcakend","BE": "SDE Backend"}
df["role_clean"]=df["role"].replace(role_mapping)
```

```python
df["current_ctc"]=df["current_ctc"].str.replace("₹","")
df["current_ctc"]=df["current_ctc"].str.replace("LPA","")
df["current_ctc"]=df["current_ctc"].str.replace(",","")
df["current_ctc"]=df["current_ctc"].str.strip()
df["current_ctc"]=pd.to_numeric(df["current_ctc"],errors="coerce")
df["current_ctc"]=df["current_ctc"].apply(lambda x:x/100000 if x > 1000 else x)
print(df["current_ctc"].head())

```

# TASK -03

```python
print(df["education_tier"].unique())
df["education_tier_clean"] = df["education_tier"].replace({
    "T1": "Tier 1",
    "1": "Tier 1",
    "Tier-1": "Tier 1",

    "T2": "Tier 2",
    "2": "Tier 2",
    "Tier-2": "Tier 2",

    "T3": "Tier 3",
    "3": "Tier 3",
    "Tier-3": "Tier 3"
})

print(df["education_tier_clean"].value_counts())
```

```python
print(df["company_type"].unique())
df["company_type"] = df["company_type"].str.strip().str.lower()
df["company_type_clean"] = df["company_type"].replace({
    "unicorn": "Unicorn",
    "mnc": "MNC",
    "mid-size": "Mid-size",
    "early-stage": "Early-stage"
})

print(df["company_type_clean"].value_counts())
```

```python
df = df.drop_duplicates()

print("Duplicate rows:", df.duplicated().sum())
```

```python
print(df.isnull().sum())
```

```python
print(df.dtypes)

print(df["role_clean"].value_counts())

print(df["company_type_clean"].value_counts())

print(df["education_tier_clean"].value_counts())
```

```python
role_salary = df.groupby("role_clean")["current_ctc"].agg(["median", "mean", "min", "max"])
role_salary = role_salary.sort_values(by="median", ascending=False)
print(role_salary)
```



The highest median CTC is for Product Manager at 31.60 LPA.

The lowest median CTC is for DA at 15.90 LPA.



```python
backend = df[df["role_clean"] == "SDE Backend"]
backend["exp_band"] = pd.cut(
    backend["years_exp"],
    bins=[0, 1, 3, 5, 100],
    labels=["0-1", "2-3", "4-5", "6+"],
    include_lowest=True
)
exp_salary = backend.groupby("exp_band")["current_ctc"].median()

print(exp_salary)
```



The median CTC for SDE Backend increases with experience.

0–1 years: 11.70 LPA

2–3 years: 20.75 LPA

4–5 years: 25.50 LPA

6+ years: 36.20 LPA

This shows that employees with more experience receive higher salaries.

```python
sde = df[df["role_clean"].isin(["SDE Backend", "SDE Frontend", "SDE Full-Stack"])]

skills = ["AWS", "ML", "System Design", "Kubernetes"]

for skill in skills:

    with_skill = sde[sde["skills"].str.contains(skill, case=False, na=False)]["current_ctc"].median()

    without_skill = sde[~sde["skills"].str.contains(skill, case=False, na=False)]["current_ctc"].median()

    print(skill)
    print("With Skill :", with_skill)
    print("Without Skill :", without_skill)
    print()
```



Among the selected skills, System Design has the highest median CTC (27.15 LPA), followed by AWS (25.90 LPA).

ML also shows a higher median CTC (22.60 LPA) than employees without ML (20.90 LPA).

Based on this dataset, learning skills like System Design and AWS may help SDEs achieve higher salaries.

```python
backend = df[df["role_clean"] == "SDE Backend"]
company_salary = backend.groupby("company_type_clean")["current_ctc"].median()
print(company_salary)
```



Among SDE Backend employees, Unicorn companies have the highest median CTC (23.45 LPA).

Mid-size companies offer a median CTC of 21.95 LPA, followed by MNCs (19.50 LPA).

Early-stage companies have the lowest median CTC at 17.15 LPA.

```python
df["exp_band"] = pd.cut(
    df["years_exp"],
    bins=[0,1,3,5,100],
    labels=["0-1","2-3","4-5","6+"],
    include_lowest=True
)
df["group_median"] = df.groupby(
    ["role_clean","company_type_clean","exp_band"]
)["current_ctc"].transform("median")
df["gap"] = df["current_ctc"] - df["group_median"]
underpaid = df.sort_values("gap").head(10)
print(underpaid[["employee_id","role_clean","company_type_clean","years_exp","current_ctc","group_median","gap"]])
```
